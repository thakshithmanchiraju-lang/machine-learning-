```python
#postlab
import seaborn as sns

# Load the Titanic dataset from seaborn
df = sns.load_dataset('titanic')
display(df.head())
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
# df = pd.read_csv("titanic.csv")   # replace with your dataset path (This line is commented out as 'df' is already loaded from seaborn)

# Basic inspection
print("Shape:", df.shape)
print("\nInfo:")
print(df.info())
print("\nSummary Stats:")
print(df.describe())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap 24eu02028")
plt.show()

# Univariate analysis
sns.countplot(x="survived", data=df)
plt.title("Survival Count 24eu02028")
plt.show()

sns.countplot(x="pclass", data=df)
plt.title("Passenger Class Distribution 24eu02028")
plt.show()

sns.countplot(x="sex", data=df)
plt.title("Gender Distribution 24eu02028")
plt.show()

# Bivariate analysis
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival by Gender 24eu02028")
plt.show()

sns.countplot(x="pclass", hue="survived", data=df)
plt.title("Survival by Class 24eu02028")
plt.show()

sns.histplot(df[df["survived"]==1]["age"], kde=True, color="green", label="Survived")
sns.histplot(df[df["survived"]==0]["age"], kde=True, color="red", label="Not Survived")
plt.legend()
plt.title("Age Distribution by Survival 24eu02028")
plt.show()

# Correlation heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap 24eu02028")
plt.show()
```
